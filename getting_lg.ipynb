```python
import skrutable.utils as _utils

from skrutable.meter_identification import MeterIdentifier, flush_profiling_report

MI = MeterIdentifier()

def get_lg(verse):
    """
    Example Use: 
    verse = "ब्रह्मादयः सुरगणा मुनयोऽथ सिद्धाः सत्त्वैकतानमतयो वचसां प्रवाहैः नाराधितुं पुरुगुणैरधुनापि पित्रः किं तोष्टुमर्हति स मे हरिरुग्रजातेः ॥"
    print(get_lg(verse))
    
    >>>
    gglglllgllglgg
    gglglllgllglgg
    gglglllgllglgg
    gglglllgllglgg
    
    The function returns the lg pattern with \n included as shown below
    'gglglllgllglgg\ngglglllgllglgg\ngglglllgllglgg\ngglglllgllglgg'
    
    """
    meter_info = MI.identify_meter(verse, resplit_option="resplit_lite", resplit_keep_midpoint=True, from_scheme="DEV")
    return meter_info.syllable_weights
```
