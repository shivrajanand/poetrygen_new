```python
df['lg-pattern'] = ""

for i, r in tqdm(df.iterrows(), total=df.shape[0]):
    input_string = r['sa']
    verse = MI.identify_meter(
        input_string,
        from_scheme='DEV',
        resplit_option='resplit_lite'
    )

    summary = verse.summarize()

    x = ""
    for line in summary.split('\n')[0:4]:
        x += line[:8]

    if len(x) == 32:
        df.at[i, 'lg-pattern'] = x
    else:
        df.at[i, 'lg-pattern'] = "ERROR\n" + x
```

```python
counts = df['lg-pattern'].value_counts().head(10)
percent = df['lg-pattern'].value_counts(normalize=True).head(10) * 100

result = pd.DataFrame({
    'count': counts,
    'percent': percent
})

result
```

```python
pattern_counts = df['lg-pattern'].value_counts()

result = (
    pattern_counts
    .value_counts()
    .sort_index()
    .rename_axis('n')
    .reset_index(name='number_of_patterns')
)

result['number_of_samples'] = (
    result['n'] * result['number_of_patterns']
)

result['percentage_of_samples'] = (
    result['number_of_samples'] / len(df) * 100
)

result
```

The table shows the **frequency distribution of metrical patterns** in the dataset. Most patterns are rare: **64.6% of samples have a unique pattern**, while **20.3% belong to patterns occurring twice**. Overall, about **84.9% of samples belong to patterns occurring no more than twice**, whereas highly recurrent patterns are very uncommon. The most frequent individual pattern occurs only **11 times**.
