```python
from skrutable.meter_identification import MeterIdentifier
import re
MI = MeterIdentifier()


import re

def extract_lg_pattern(verse, verbose=False):
    if verbose: print(verse)
    verse = MI.identify_meter(
            verse,
            from_scheme='DEV',
            resplit_option='resplit_lite'
        )
    
    summary = verse.summarize()
    if verbose: print(summary)

    patterns = re.findall(
        r'^([lg]+)\s+\{m:',
        summary,
        re.MULTILINE
    )[:4]

    return ''.join(patterns)
```

```python
for i, r in tqdm(df.iterrows(), total=df.shape[0]):
    verse = r['sa']
    pattern = extract_lg_pattern(verse)
    
    if len(pattern) == r['syllable_count']:
        continue
    else:
        print(
            f"Error at row {i}. "
            f"Meter = {r['meter_cd']}. "
            f"Pattern length = {len(pattern)}, "
            f"Expected = {r['syllable_count']}. "
            f"Pattern = {pattern}"
        )
```
